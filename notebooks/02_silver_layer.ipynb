### silver layer for customer bronze table


```python
# Read Bronze customers table
df_customers = spark.table("bronze_customers")

# Display the data
display(df_customers)
```

#### Cell 2 — Clean customers

```python
from pyspark.sql.functions import col, trim, to_date

df_silver_customers = (
    df_customers
    .dropDuplicates(["customer_id"])
    .filter(col("customer_id").isNotNull())
    .withColumn("customer_id", trim(col("customer_id")))
    .withColumn("name", trim(col("name")))
    .withColumn("city", trim(col("city")))
    .withColumn("state", trim(col("state")))
    .withColumn(
        "registration_date",
        to_date(col("registration_date"), "yyyy-MM-dd")
    )
)

display(df_silver_customers)
```

#### Cell 3 — Write Silver Customers and Verify

```python
df_silver_customers.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("silver_customers")

df_check = spark.table("silver_customers")

display(df_check)
```

### Cell 5 — Read Bronze Accounts


```python
df_accounts = spark.table("bronze_accounts")

display(df_accounts)
```

#### Cell 6 — Clean Silver Accounts

```python
from pyspark.sql.functions import col, trim, to_date

df_silver_accounts = (
    df_accounts
    .dropDuplicates(["account_id"])
    .filter(col("account_id").isNotNull())
    .filter(col("customer_id").isNotNull())
    .withColumn("account_id", trim(col("account_id")))
    .withColumn("customer_id", trim(col("customer_id")))
    .withColumn("account_type", trim(col("account_type")))
    .withColumn(
        "opening_date",
        to_date(col("opening_date"), "yyyy-MM-dd")
    )
    .withColumn("balance", col("balance").cast("double"))
)

display(df_silver_accounts)
```

#### Cell 7 — Write Silver Accounts and Verify

```python
df_silver_accounts.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("silver_accounts")

display(spark.table("silver_accounts"))
```

### Cell 9 — Read Bronze Transactions
