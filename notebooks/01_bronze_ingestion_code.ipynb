### customer file ingestion--------------------------

```python

DROP TABLE IF EXISTS bronze_customers;

CREATE TABLE bronze_customers
USING DELTA
AS
SELECT
    CAST(NULL AS STRING) AS customer_id,
    CAST(NULL AS STRING) AS name,
    CAST(NULL AS STRING) AS city,
    CAST(NULL AS STRING) AS state,
    CAST(NULL AS STRING) AS registration_date
WHERE 1 = 0;

COPY INTO bronze_customers
FROM 'abfss://raw@datalakebankinganalytics.dfs.core.windows.net/'
WITH (
  CREDENTIAL (AZURE_SAS_TOKEN = 'sp=rl&st=2026-10-02T15:41:03Z&se=2026-10-02T23:56:03Z&spr=https&sv=2026-02-06&sr=c&sig=yxl4it1MnWxVgybA9XzhQXf2274LTwbShS2LqY%2FVxe4%3D')
)
FILEFORMAT = CSV
FILES = ('customers.csv')
FORMAT_OPTIONS ('header' = 'true');

SELECT * FROM bronze_customers LIMIT 10;

```
